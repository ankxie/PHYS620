Consider the mass confined to the surface of a cone described in Taylor example 13.4 (page 533). There are maximum and minimum heights $z_{max}$ and $z_{min}$ between which the mass is confined. Plot the function $H(z)$ for $z > 0$ and $z \rightarrow \infty$. In a markdown cell, describe the motion indicated by your plot.

Hint: At both extrema, $\dot{z} = 0$. Fix the value of vertical momentum to $p_z = 0$ to make your plot. 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 1 #arbitrary values for the constants
c = 1
p_phi = 1
g = 10

z_array = np.linspace(0, 10, 100)
H = (p_phi**2) * (2*m*c**2 *z_array**2)**-1 + m * g * z_array

fig, ax = plt.subplots()

ax.plot(z_array, H, label = r'$H_{extrema}$')
ax.plot(z_array, np.full_like(z_array, 40), ls = '--', c = 'k', label = r'$E$')

ax.set_ylabel(r'$H(z)$')
ax.set_xlabel(r'$z$')
ax.legend()
```

This figure shows the values of $H$ as a function of $z$ for when that value of $z$ is an extrema. If the system has constant total energy (and therefore a constant Hamiltonian), it will have two extrema. The mass's motion will be confined between these two extrema. There is a minimum value for the total energy that the system can have, and at this energy, the mass will be confined to a single value of z.
